**Спринт ревью №0**

Ваша задача — разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Вам нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм. 
1. Класс Растение (Plant) 
Поля (свойства): 
— Имя (string): название растения (например, «Роза»). 
— Тип (string): тип растения (например, «Цветок» или «Дерево»). 
— Высота (float): высота растения в сантиметрах. 
Методы: 
— Цветение(): метод, который выводит сообщение о том, что растение цветет. 
— Растет(float рост): метод для увеличения высоты растения на заданное значение. 

2. Класс Цветок (Flower) — наследуется от класса Растение 
Поля (свойства): 
— Цвет (string): цвет цветка (например, «Красный»). 
— КоличествоЛепестков (int): количество лепестков у цветка. 
Методы: 
— Цветение(): переопределить метод из класса Растение, чтобы вывести специальное сообщение для цветков. 
— Увядать(): метод, который сообщает, что цветок начинает увядать. 

3. Класс Дерево (Tree) — наследуется от класса Растение 
Поля (свойства): 
— Возраст (int): возраст дерева в годах. 
— ТипЛистьев (string): тип листьев (например, «Хвойные» или «Лиственные»). 
Методы: 
— Цветение(): переопределить метод из класса Растение с сообщением, специфичным для дерева. 
— Вырастить(int годы): метод, который увеличивает возраст дерева и сообщает о новом возрасте. 
Полиморфизм 
Каждый класс (цветок и дерево) должен использовать свой вариант метода Цветение(), чтобы продемонстрировать полиморфизм. Например: 
— При вызове Цветение() для цветка должно выводиться: "Цветок [Имя] распускается и радует глаз!" 
— При вызове Цветение() для дерева должно выводиться: "Дерево [Имя] покрывается новыми листьями!" 

Задание: 
1. Создайте объекты для одного цветка и одного дерева. 
2. Вызовите методы, чтобы показать, как растения цветут и растут. 
3. Реализуйте полиморфизм через переопределение метода Цветение() в классах-наследниках.

```
using System;

public class Plant
{
    public string Имя { get; set; }
    public string Тип { get; set; }
    public float Высота { get; set; }

    public Plant(string имя, string тип, float высота)
    {
        Имя = имя;
        Тип = тип;
        Высота = высота;
    }

    public virtual void Цветение()
    {
        Console.WriteLine($"Растение {Имя} цветет.");
    }

    public void Растет(float рост)
    {
        Высота += рост;
        Console.WriteLine($"Растение {Имя} выросло на {рост} см. Текущая высота: {Высота} см.");
    }
}

public class Flower : Plant
{
    public string Цвет { get; set; }
    public int КоличествоЛепестков { get; set; }

    public Flower(string имя, float высота, string цвет, int количествоЛепестков)
        : base(имя, "Цветок", высота)
    {
        Цвет = цвет;
        КоличествоЛепестков = количествоЛепестков;
    }

    public override void Цветение()
    {
        Console.WriteLine($"Цветок {Имя} распускается и радует глаз!");
    }

    public void Увядать()
    {
        Console.WriteLine($"Цветок {Имя} начинает увядать...");
    }
}

public class Tree : Plant
{
    public int Возраст { get; set; }
    public string ТипЛистьев { get; set; }

    public Tree(string имя, float высота, int возраст, string типЛистьев)
        : base(имя, "Дерево", высота)
    {
        Возраст = возраст;
        ТипЛистьев = типЛистьев;
    }

    public override void Цветение()
    {
        Console.WriteLine($"Дерево {Имя} покрывается новыми листьями!");
    }

    public void Вырастить(int годы)
    {
        Возраст += годы;
        Console.WriteLine($"Дерево {Имя} выросло на {годы} лет. Новый возраст: {Возраст} лет.");
    }
}

Flower rose = new Flower("Роза", 25.5f, "Красный", 32);
Tree oak = new Tree("Дуб", 150.0f, 5, "Лиственные");

Console.WriteLine("ЦВЕТОК");
rose.Цветение();
rose.Растет(5.5f);
rose.Увядать();

Console.WriteLine();

Console.WriteLine("ДЕРЕВО");
oak.Цветение();
oak.Растет(30.0f);
oak.Вырастить(2);

Console.WriteLine();

Console.WriteLine("ПОЛИМОРФИЗМ");
Plant[] plants = { rose, oak };

foreach (Plant plant in plants)
{
    plant.Цветение();
}
```
