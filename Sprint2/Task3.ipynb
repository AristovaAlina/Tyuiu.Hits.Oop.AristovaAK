Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.
Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IResearcher
{
    void DoResearch();
}

public interface ITeacher
{
    void Teach(string subject);
}


public class Student
{
    private string _name;
    private int _age;
    private int _course;
    private double _gpa;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set { _age = value; }
    }

    public int Course
    {
        get { return _course; }
        set { _course = value; }
    }

    public double GPA
    {
        get { return _gpa; }
        set { _gpa = value; }
    }

    
    public Student() // конструктор по умолчанию
    {
        Name = "Любое";
        Age = 18;
        Course = 1;
        GPA = 0.0;
    }

    public Student(string name, int age, int course, double gpa) // конструктор с параметрами
    {
        Name = name;
        Age = age;
        Course = course;
        GPA = gpa;
    }


    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Курс: {Course}, GPA: {GPA}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен.");
    }

    public virtual void EnrollCourse(string courseName)
    {
        Console.WriteLine($"{Name} записался на курс: {courseName}");
    }


    public void Help(Student other) // взаимодействие объектов
    {
        Console.WriteLine($"{Name} помогает {other.Name} с учёбой.");
    }
}


public class BachelorStudent : Student // простое наследование
{
    private string _specialty;

    public string Specialty
    {
        get { return _specialty; }
        set { _specialty = value; }
    }

    public BachelorStudent() : base()
    {
        Specialty = "Не определена";
    }

    public BachelorStudent(string name, int age, int course, double gpa, string specialty)
        : base(name, age, course, gpa)
    {
        Specialty = specialty;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности {Specialty}.");
    }

    public void WriteCourseWork(string topic)
    {
        Console.WriteLine($"{Name} пишет курсовую на тему: {topic}");
    }
}


public class MasterStudent : Student, IResearcher // сложное и множественное наследование
{
    private string _scientificAdvisor;

    public string ScientificAdvisor
    {
        get { return _scientificAdvisor; }
        set { _scientificAdvisor = value; }
    }

    public MasterStudent() : base()
    {
        ScientificAdvisor = "Не назначен";
    }

    public MasterStudent(string name, int age, int course, double gpa, string advisor)
        : base(name, age, course, gpa)
    {
        ScientificAdvisor = advisor;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}";
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен. Руководитель: {ScientificAdvisor}.");
    }

    public void DoResearch()
    {
        Console.WriteLine($"{Name} проводит исследование под руководством {ScientificAdvisor}.");
    }
}


public class PhDStudent : MasterStudent, ITeacher //сложное и множественное наследование
{
    private string _dissertationTopic;

    public string DissertationTopic
    {
        get { return _dissertationTopic; }
        set { _dissertationTopic = value; }
    }

    public PhDStudent() : base()
    {
        DissertationTopic = "Не определена";
    }

    public PhDStudent(string name, int age, int course, double gpa,
                      string advisor, string topic)
        : base(name, age, course, gpa, advisor)
    {
        DissertationTopic = topic;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {DissertationTopic}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает материалы по теме: {DissertationTopic}.");
    }

    public void Teach(string subject)
    {
        Console.WriteLine($"{Name} ведёт занятия по предмету: {subject}");
    }
}


var bachelor = new BachelorStudent("Анна", 19, 2, 4.7, "Программная инженерия");
var master   = new MasterStudent("Иван", 23, 1, 4.9, "доц. Петров");
var phd      = new PhDStudent("Мария", 26, 2, 5.0, "проф. Смирнова", "Разработка ПО для расчёта прочности");

Console.WriteLine("Простое:");
Console.WriteLine(bachelor.GetInfo());
bachelor.Study();
bachelor.WriteCourseWork("Веб-приложение");

Console.WriteLine("\nСложное:");
Console.WriteLine(phd.GetInfo());
phd.TakeExam();      // переопределён в MasterStudent
phd.DoResearch();    // унаследован от MasterStudent
phd.Teach("Алгоритмы"); // реализация ITeacher

Console.WriteLine("\nМножественное (через интерфейсы):");
Student[] all = { bachelor, master, phd };
foreach (var s in all)
{
    if (s is IResearcher r)
        r.DoResearch();
    if (s is ITeacher t)
        t.Teach("Математика");
}

Console.WriteLine("\nПолиморфизм:");
foreach (var s in all)
{
    Console.WriteLine(s.GetInfo());
    s.Study();
    Console.WriteLine(new string('-', 40));
}

Console.WriteLine("\nВзаимодействие");
bachelor.Help(master);
master.Help(phd);
```
