<h2 style="color:DodgerBlue">Реализация:</h2>
Тема: «Наследование, Абстрактный класс и интерфейсы»
Цель задания: создать классы, представляющие структуру коммерческого предприятия, используя простое, сложное и множественное наследование.
Условия задания:
1.	Создание базового класса:
•	Создайте базовый класс Employee (Сотрудник) с полями:
•	Name (строка) - имя сотрудника
•	Position (строка) - должность сотрудника
•	Salary (число с плавающей точкой) - зарплата сотрудника
•	Реализуйте метод PrintInfo(), который выводит информацию о сотруднике на консоль.
2.	Простое наследование:
•	Создайте класс Manager (Менеджер), который наследует от Employee.
•	Добавьте поле Department (строка) для указания отдела, которым управляет менеджер.
•	Переопределите метод PrintInfo(), чтобы он также выводил информацию о department.
3.	Сложное наследование:
•	Создайте класс ProjectManager (Проектный менеджер), который наследует от Manager.
•	Добавьте поле ProjectName (строка) для указания названия проекта, за который отвечает проектный менеджер.
•	Переопределите метод PrintInfo(), чтобы он также выводил название проекта.
4.	Множественное наследование:
•	В C# нельзя реализовать множественное наследование напрямую, но вы можете использовать интерфейсы.
•	Создайте интерфейсы:
•	IReportable с методом GenerateReport(), который возвращает строку с отчетом.
•	ITeamLeader с методом GetTeamMembers(), который возвращает список членов команды (например, массив строк).
•	Реализуйте проектный менеджер, который будет реализовывать оба интерфейса.
5.	Тестирование:
•	Создайте экземпляры каждого класса (Employee, Manager, ProjectManager).
•	Вызовите метод PrintInfo() для каждого экземпляра.
•	Реализуйте методы интерфейсов в ProjectManager и протестируйте их.

```
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo() // virtual — чтобы наследники могли переопределить метод
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:C}");
    }
}


public class Manager : Employee // простое наследование
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}


public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}


public class ProjectManager : Manager, IReportable, ITeamLeader // сложное и множественное наследование
{
    public string ProjectName { get; set; }
    private string[] _teamMembers;

    public ProjectManager(string name, string position, double salary,
                          string department, string projectName, string[] teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        _teamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }


    public string GenerateReport()
    {
        return $"Отчёт по проекту \"{ProjectName}\":\n" +
               $"  Руководитель: {Name}\n" +
               $"  Отдел: {Department}\n" +
               $"  Команда ({_teamMembers.Length} чел.): {string.Join(", ", _teamMembers)}";
    }

    public string[] GetTeamMembers()
    {
        return _teamMembers;
    }
}


string[] team = { "Александр Пушкин", "Агата Кристи", "Антон Чехов" };

Console.WriteLine("Обычный сотрудник");
var employee = new Employee("Александр Пушкин", "Разработчик", 70000);
employee.PrintInfo();

Console.WriteLine("\nМенеджер");
var manager = new Manager("Агата Кристи", "Менеджер", 100000, "Отдел разработки");
manager.PrintInfo();

Console.WriteLine("\nПроектный менеджер");
var projectManager = new ProjectManager("Антон Чехов", "Проектный менеджер", 130000, "Отдел разработки", "Мобильное приложение", team);
projectManager.PrintInfo();


Console.WriteLine("\nОтчёт:");
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("\nЧлены команды:");
foreach (var member in projectManager.GetTeamMembers())
{
    Console.WriteLine($" - {member}");
}
```
