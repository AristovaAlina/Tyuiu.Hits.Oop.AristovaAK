```
// ========== ТОВАР ==========
public class Product
{
    // Приватные поля
    private string name;
    private decimal price;
    private int quantity;

    // Защищённое поле — доступно производным классам
    protected string category;

    // Внутреннее поле — доступно внутри сборки
    internal string description;

    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    // Конструктор по умолчанию
    public Product()
    {
        Name = "Без названия";
        Price = 0;
        Quantity = 0;
        Category = "Общее";
        Description = "Нет описания";
    }

    // Конструктор с параметрами
    public Product(string name, decimal price, int quantity,
                   string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    // Общая стоимость всех единиц товара
    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    // Обновление количества
    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
        if (Quantity < 0) Quantity = 0;
    }

    // Полное описание товара
    public string GetProductInfo()
    {
        return $"Товар: {Name} | Категория: {Category} | Цена: {Price} руб. | " +
               $"Количество: {Quantity} | Описание: {Description}";
    }
}

// ========== ПРОДАВЕЦ ==========
public class Seller
{
    // Приватные поля
    private string name;
    private string employeeId;

    // Защищённое поле
    protected decimal salary;

    // Внутреннее поле
    internal string contactInfo;

    // Список товаров продавца
    private List<Product> products = new List<Product>();

    // Свойства
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    // Конструктор по умолчанию
    public Seller()
    {
        Name = "Безымянный";
        EmployeeId = "0000";
        Salary = 0;
        ContactInfo = "нет контактов";
    }

    // Конструктор с параметрами
    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    // Добавить товар продавцу
    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар \"{product.Name}\" добавлен продавцу {Name}.");
    }

    // Продать товар — уменьшить количество
    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продано {quantity} шт. товара \"{product.Name}\". " +
                              $"Осталось: {product.Quantity}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара \"{product.Name}\" на складе.");
        }
    }

    // Информация о продавце
    public string GetSellerInfo()
    {
        return $"Продавец: {Name} | ID: {EmployeeId} | Зарплата: {Salary} руб. | " +
               $"Контакты: {ContactInfo}";
    }

    // Список товаров продавца
    public List<Product> GetProducts()
    {
        return products;
    }
}

// ========== МАГАЗИН ==========
public class Store
{
    // Приватные поля
    private string storeName;
    private string location;

    // Публичное поле
    public string storeHours;

    // Список продавцов
    private List<Seller> sellers = new List<Seller>();

    // Свойства
    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    // Конструктор по умолчанию
    public Store()
    {
        StoreName = "Без названия";
        Location = "Не указано";
        storeHours = "10:00 - 20:00";
    }

    // Конструктор с параметрами
    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    // Добавить продавца
    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} принят на работу в {StoreName}.");
    }

    // Перечислить все товары магазина
    public void ListProducts()
    {
        Console.WriteLine($"\n--- Товары в магазине \"{StoreName}\" ---");
        foreach (var seller in sellers)
        {
            foreach (var product in seller.GetProducts())
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }

    // Информация о магазине
    public string GetStoreInfo()
    {
        return $"Магазин: {StoreName} | Адрес: {Location} | Часы работы: {storeHours} | " +
               $"Продавцов: {sellers.Count}";
    }
}
```

```
var toy = new Product("Плюшевый мишка", 1500m, 10, "Мягкие игрушки", "Большой коричневый медведь");

var seller = new Seller("Анна", "EMP-001", 45000m, "anna@toyshop.ru");

var store = new Store("Мир игрушек", "г. Москва, ул. Ленина, 15", "09:00 - 21:00");

// ========== ВЗАИМОДЕЙСТВИЕ ==========

store.AddSeller(seller);
seller.AddProduct(toy);

Console.WriteLine();
Console.WriteLine(toy.GetProductInfo());
Console.WriteLine($"Общая стоимость: {toy.GetTotalPrice()} руб.");

Console.WriteLine();
seller.SellProduct(toy, 3);
Console.WriteLine($"Общая стоимость после продажи: {toy.GetTotalPrice()} руб.");

Console.WriteLine();
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(store.GetStoreInfo());

store.ListProducts();
```
